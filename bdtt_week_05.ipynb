```sql
%python
from pyspark.sql.types import *
mySchema = StructType([
    StructField('index', IntegerType()),
    StructField('webpage', StringType()),
    StructField('associated_files', StringType())
])

webpages1 = spark.read.options(delimiter="\t").schema(mySchema).csv(f'{week_4_directory}/webpage/')
```

```sql
%python
webpages1.show(5)
```

```sql
%python
webpages1.createOrReplaceTempView('sqlwebpages1')
```

```sql
SELECT * FROM sqlwebpages1 LIMIT 10
```

```sql
SHOW DATABASES
```

```sql
SHOW TABLES
```

```sql
CREATE OR REPLACE TABLE default.webtable AS SELECT * FROM sqlwebpages1
```

```sql
SHOW TABLES
```

```sql
CREATE DATABASE IF NOT EXISTS bdtt_db;
SHOW DATABASES
```

```sql
SHOW TABLES IN default;
```

```sql
-- DROP TABLE IF EXISTS bdtt_db.webtable;
```

```sql
SHOW TABLES IN bdtt_db;
```

```sql
CREATE OR REPLACE TABLE bdtt_db.webtable AS SELECT * FROM default.webtable;
```

```sql
SHOW TABLES IN bdtt_db;
```

```sql
CREATE OR REPLACE VIEW bdtt_db.s_webtable as SELECT * FROM bdtt_db.webtable WHERE webpage LIKE 's%';
```

```sql
SELECT * FROM bdtt_db.s_webtable LIMIT 5;
```

```sql
SELECT COUNT(*) FROM bdtt_db.webtable
```

```sql
SHOW TABLES IN bdtt_db
```

```sql
CREATE OR REPLACE TEMP VIEW sub_webpages AS SELECT webpage, associated_files FROM bdtt_db.webtable;
```

```sql
SELECT * FROM sub_webpages LIMIT 5;
-- CREATE OR REPLACE GLOBAL TEMP VIEW sub_webpages AS SELECT webpage, associated_files FROM bdtt_db.webtable;
```

```sql
CREATE OR REPLACE TEMP VIEW spliteg AS SELECT split(webpage, '_') as split_wp, associated_files FROM bdtt_db.webtable
```

```sql
SELECT * FROM spliteg LIMIT 10;
```

```sql
CREATE OR REPLACE TEMP VIEW splitwebpages AS SELECT webpage, split(associated_files, ",") AS split_af FROM bdtt_db.webtable;
```

```sql
SELECT * FROM splitwebpages LIMIT 10;
```

```sql
CREATE OR REPLACE TEMP VIEW exploded_webpages AS SELECT webpage, explode(split_af) FROM splitwebpages
```

```sql
SELECT * FROM exploded_webpages LIMIT 10;
```

```sql
SELECT COUNT(*) FROM exploded_webpages WHERE col LIKE '%.jpg';
```

```sql
SELECT COUNT(*) FROM exploded_webpages WHERE col LIKE '%.css';
```

```sql
SELECT COUNT(*) FROM exploded_webpages WHERE col LIKE '%.js';
```

```sql
CREATE OR REPLACE TEMP VIEW web_js AS SELECT webpage, col FROM exploded_webpages WHERE col LIKE '%.js';
```

```sql
SELECT * FROM web_js LIMIT 10;
```
