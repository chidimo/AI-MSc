````python
import os
import tarfile
import requests
from dotenv import load_dotenv

load_dotenv()
````

````python
# 🔑 Read API key from environment
API_KEY = os.getenv("MOZILLA_COLLECTIVE_API_KEY")

if not API_KEY:
    raise ValueError("MOZILLA_COLLECTIVE_API_KEY not set in environment variables")
````

````python

# 📦 Configure dataset
DATASET_ID = "cmll4w0ln00nemn07axe7mj09"

OUTPUT_FILE = "dataset.tar.gz"
BASE_DIR = os.path.join(os.getcwd(), "cia_factbook")
OUTPUT_PATH = os.path.join(BASE_DIR, OUTPUT_FILE)

BASE_URL = "https://datacollective.mozillafoundation.org/api/datasets"


def get_download_url(api_key: str, dataset_id: str) -> str:
    url = f"{BASE_URL}/{dataset_id}/download"

    headers = {
        "Authorization": f"Bearer {api_key}",
        "Content-Type": "application/json",
    }

    response = requests.post(url, headers=headers)
    response.raise_for_status()

    return response.json().get("downloadUrl")


def download_file(url: str, output_path: str):
    os.makedirs(os.path.dirname(output_path), exist_ok=True)
    with requests.get(url, stream=True) as r:
        r.raise_for_status()
        with open(output_path, "wb") as f:
            for chunk in r.iter_content(chunk_size=8192):
                if chunk:
                    f.write(chunk)

def extract_tar_gz(file_path: str, extract_to: str = "."):
    if not tarfile.is_tarfile(file_path):
        raise ValueError("Not a valid tar.gz file")

    with tarfile.open(file_path, "r:gz") as tar:
        tar.extractall(path=extract_to)

    print(f"Extracted to: {os.path.abspath(extract_to)}")
````

````python
## 🚀 Run download
download_url = get_download_url(API_KEY, DATASET_ID)
if not download_url:
    raise ValueError("Failed to retrieve download URL")

print(f"Downloading from: {download_url}")

download_file(download_url, OUTPUT_PATH)

print(f"Download complete: {OUTPUT_PATH}")

extract_tar_gz(OUTPUT_PATH, BASE_DIR)
````
