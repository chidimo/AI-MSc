```python
# Challenge 4

def count_lines(f):
    with open(f, "r") as fp:
        return len(fp.readlines())
print('Total lines: ', count_lines("/Volumes/teaching/datasets/raw_data/week_1/activity_log.txt"))

def count_lines_with_text(f, text):
    with open(f, "r") as fp:
        return len([line for line in fp.readlines() if text in line])
count_lines_with_text("/Volumes/teaching/datasets/raw_data/week_1/activity_log.txt", "outcome=FAIL")

percentage_failed = (64/2500) * 100
print('Percentage of failed runs: ', percentage_failed, '%')
```

```python
# Challenge 5
def words_per_line(f):
    with open(f, "r") as fp:
        lines = fp.readlines()
        total_lines = len(lines)
        total_words = sum([len(line.split()) for line in lines])
    print('total words: ', total_words)
    print('total lines', total_lines)
    return total_words / total_lines

words_per_line("/Volumes/teaching/datasets/raw_data/week_2/purplecow.txt")
```

```python
# Challenge 6

def length_of_longest_line(f):
    with open(f, "r") as fp:
        lines = fp.readlines()
        longest_line = max(lines, key=len)
    return longest_line
length_of_longest_line("/Volumes/teaching/datasets/raw_data/week_2/test.json")
```

```python
# Challenge 7
def print_line_with_word_occurrence(f, word):
    with open(f, "r") as fp:
        for line in fp.readlines():
            print(line)
            if word in line:
                print(line)
print_line_with_word_occurrence("/Volumes/teaching/datasets/raw_data/week_2/purplecow.txt", "purple")
```
