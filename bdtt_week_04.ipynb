```python
webpages = spark.read.options(delimiter="\t").csv(f"{base_directory}/webpage/*")
```

```python
webpages.show(5, truncate=False)
```

```python
webpages.printSchema()
```

```python
webpages = (
    webpages.withColumnRenamed("_c0", "index")
    .withColumnRenamed("_c1", "webpage")
    .withColumnRenamed("_c2", "associated_files")
)

webpages.show(5, truncate=False)
```

```python
webpages = webpages.withColumn("index", webpages.index.cast("int"))
webpages.printSchema()
webpages.show(5, truncate=False)
```

```python
from pyspark.sql import functions as F
```

```python
webpagesDF = webpages.select('webpage', 'associated_files')
```

```python
step2DF = webpagesDF.withColumn('associated_files_array', F.split(F.col('associated_files'), ","))
```

```python
step2DF.show(5, truncate=False)
```

```python
step3DF = step2DF.withColumn(
    'associated_file',
    F.explode(F.col('associated_files_array'))
)

step3DF.show(5, truncate=False)
```
